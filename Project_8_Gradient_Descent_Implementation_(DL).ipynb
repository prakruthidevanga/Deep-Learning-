## Step 1: Import Required Libraries

We use NumPy for numerical calculations and Matplotlib for visualizing the loss and predictions.

```python
import numpy as np
import matplotlib.pyplot as plt

print("Libraries imported successfully!")
```

## Step 2: Create a Simple Dataset

We create a small dataset where the relationship between X and y is approximately linear.

The expected relationship is:

y = 2x + 1

The model will learn the values of the slope and intercept using Gradient Descent.

```python
# Input data
X = np.array([1, 2, 3, 4, 5], dtype=float)

# Target/output data
y = np.array([3, 5, 7, 9, 11], dtype=float)

print("X values:", X)
print("y values:", y)
```

## Step 3: Visualize the Dataset

The following graph shows the relationship between the input values and target values.

```python
plt.figure(figsize=(7, 5))

plt.scatter(X, y)

plt.xlabel("Input X")
plt.ylabel("Target y")
plt.title("Simple Dataset")

plt.grid(True)
plt.show()
```

## Step 4: Initialize Model Parameters

We use a simple linear regression model:

y_pred = wX + b

where:

- w = weight/slope
- b = bias/intercept
- y_pred = predicted output

Initially, we set both parameters to zero.

```python
# Initialize parameters
w = 0.0
b = 0.0

# Learning rate
learning_rate = 0.01

# Number of iterations
iterations = 1000

print("Initial weight (w):", w)
print("Initial bias (b):", b)
print("Learning rate:", learning_rate)
print("Number of iterations:", iterations)
```

## Step 5: Define Prediction and Loss

The model prediction is:

y_pred = wX + b

We use Mean Squared Error (MSE) as the loss function.

MSE measures the average squared difference between the actual values and predicted values.

```python
def predict(X, w, b):
    return w * X + b


def calculate_loss(y, y_pred):
    return np.mean((y - y_pred) ** 2)


# Test the initial prediction
y_pred = predict(X, w, b)
loss = calculate_loss(y, y_pred)

print("Initial predictions:", y_pred)
print("Initial loss:", loss)
```

## Step 6: Implement Gradient Descent

Gradient Descent updates the model parameters in the direction that reduces the loss.

For the linear regression model:

y_pred = wX + b

The gradients are:

dw = (-2/n) * Σ[X(y - y_pred)]

db = (-2/n) * Σ(y - y_pred)

The parameters are updated using:

w = w - learning_rate * dw

b = b - learning_rate * db

We repeat this process for multiple iterations.

```python
# Reset parameters
w = 0.0
b = 0.0

# Store loss values
loss_history = []

# Store parameter values
w_history = []
b_history = []

n = len(X)

# Gradient Descent
for i in range(iterations):

    # Step 1: Make predictions
    y_pred = predict(X, w, b)

    # Step 2: Calculate loss
    loss = calculate_loss(y, y_pred)

    # Store values
    loss_history.append(loss)
    w_history.append(w)
    b_history.append(b)

    # Step 3: Calculate gradients
    dw = (-2 / n) * np.sum(X * (y - y_pred))
    db = (-2 / n) * np.sum(y - y_pred)

    # Step 4: Update parameters
    w = w - learning_rate * dw
    b = b - learning_rate * db

print("Gradient Descent completed!")
print("Final weight (w):", w)
print("Final bias (b):", b)
```

## Step 7: Analyze Parameter Updates

The following table shows how the model parameters change during training.

```python
print("Parameter updates during training:\n")

selected_iterations = [0, 1, 2, 5, 10, 50, 100, 500, 999]

print("Iteration\tWeight\t\tBias\t\tLoss")
print("-" * 60)

for i in selected_iterations:
    print(
        f"{i}\t\t"
        f"{w_history[i]:.6f}\t"
        f"{b_history[i]:.6f}\t"
        f"{loss_history[i]:.6f}"
    )
```

## Step 8: Visualize Loss Reduction

The loss graph shows whether the Gradient Descent algorithm is successfully learning.

If Gradient Descent works correctly, the loss should decrease as the number of iterations increases.

```python
plt.figure(figsize=(8, 5))

plt.plot(loss_history)

plt.xlabel("Iteration")
plt.ylabel("Loss (MSE)")
plt.title("Loss Reduction During Gradient Descent")

plt.grid(True)
plt.show()
```

## Step 9: Visualize Parameter Updates

We can also observe how the weight and bias change during training.

```python
plt.figure(figsize=(8, 5))

plt.plot(w_history, label="Weight (w)")
plt.plot(b_history, label="Bias (b)")

plt.xlabel("Iteration")
plt.ylabel("Parameter Value")
plt.title("Parameter Updates During Training")

plt.legend()
plt.grid(True)

plt.show()
```

## Step 10: Generate Final Predictions

After training, we use the learned weight and bias to make predictions.

```python
# Make final predictions
final_predictions = predict(X, w, b)

print("Final weight:", w)
print("Final bias:", b)

print("\nActual values:")
print(y)

print("\nPredicted values:")
print(final_predictions)
```

## Step 11: Compare Actual and Predicted Values

The following graph compares the original data points with the predictions made by the trained model.

```python
plt.figure(figsize=(8, 5))

plt.scatter(X, y, label="Actual Data")

plt.plot(X, final_predictions, label="Predicted Line")

plt.xlabel("Input X")
plt.ylabel("Output y")
plt.title("Actual vs Predicted Values")

plt.legend()
plt.grid(True)

plt.show()
```

## Step 12: Calculate Final Loss

We calculate the loss after training and compare it with the initial loss.

```python
initial_loss = loss_history[0]

final_predictions = predict(X, w, b)
final_loss = calculate_loss(y, final_predictions)

print("Initial Loss:", initial_loss)
print("Final Loss:", final_loss)

print("\nLoss Reduction:", initial_loss - final_loss)
```
